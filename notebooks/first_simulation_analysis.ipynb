# First Simulation Analysis

This notebook runs a tiny deterministic batch and summarizes outcomes, event counts, and selected action frequency. It uses the current base-game skeleton, so power, bonus-card, and round-goal scoring are intentionally not part of the results yet.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT / 'src'))
sys.path.insert(0, str(PROJECT_ROOT))
```

```python
from wingspan_ai.agents import GreedyBaselineAgent, RandomLegalAgent
from wingspan_ai.content.loader import load_base_game_content_catalog
from wingspan_ai.simulation import run_single_game
from analysis.simulation_summary import action_frequency, event_counts, outcome_rows

catalog = load_base_game_content_catalog(PROJECT_ROOT / 'wingspan-card-list.xlsx')
results = [
    run_single_game(
        catalog,
        [
            RandomLegalAgent(agent_id='random_legal_p1', random_seed=seed),
            GreedyBaselineAgent(agent_id='greedy_immediate_p2'),
        ],
        random_seed=seed,
    )
    for seed in range(1, 6)
]

rows = outcome_rows(results)
rows[:4]
```

```python
all_events = [event for result in results for event in result.events]
event_counts(all_events)
```
